## 1. Load a Transformer

We use **DistilBERT**, a  Transformer fine-tuned for positive/negative sentiment classification.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.eval()

# Look inside the Transformer architecture.
print("Transformer layers:", model.config.n_layers)
print("Attention heads per layer:", model.config.n_heads)
print("Hidden size:", model.config.dim)
```

## 2. Tokenizing the text

```python
sentence = "The movie was not good."

# Tokenization: converts text into token IDs
inputs = tokenizer(sentence, return_tensors="pt")

# [CLS] = start token, [SEP] = end token
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

print("Tokens:", tokens)
```

## 3. Embeddings and self-attention

Each token ID becomes a **vector (embedding)**. Inside each Transformer layer, **self-attention** lets a token use information from other tokens in the same sentence.

```python
import torch

# Enable attention weights
model.set_attn_implementation("eager")

with torch.no_grad():
    embeddings = model.distilbert.embeddings(inputs["input_ids"])
    outputs = model(**inputs, output_attentions=True)

# Attention from the first layer and first head
attention = outputs.attentions[0][0, 0]

print("Embeddings shape:", embeddings.shape)
print("Attention shape:", attention.shape)
print("Tokens:", tokens)
print("Attention weights:\n", attention)
```

## 4. Classification: From Transformer to prediction

The Transformer produces contextual representations. The **classification head** turns them into two scores (*logits*). **Softmax** converts these into probabilities that sum to 1.

```python
# Get prediction scores from the Transformer
with torch.no_grad():
    logits = model(**inputs).logits

# Convert scores into probabilities
probs = torch.softmax(logits, dim=1)[0]

# Display the prediction
prediction = probs.argmax().item()

print("Prediction:", model.config.id2label[prediction])
print("Confidence:", round(probs[prediction].item(), 3))
```

## Task solution

### 1–2. Load BERT and count its attention heads

`bert-base-uncased` has 12 Transformer layers with 12 attention heads per layer. The total is therefore **12 × 12 = 144 attention heads** across the model.


```python
from transformers import AutoTokenizer, AutoModel
import torch

bert_model_name = "bert-base-uncased"
bert_tokenizer = AutoTokenizer.from_pretrained(bert_model_name)
# Eager attention makes the attention weights available explicitly.
bert_model = AutoModel.from_pretrained(
    bert_model_name, attn_implementation="eager"
)
bert_model.eval()

num_layers = bert_model.config.num_hidden_layers
heads_per_layer = bert_model.config.num_attention_heads
total_heads = num_layers * heads_per_layer

print("Transformer layers:", num_layers)
print("Attention heads per layer:", heads_per_layer)
print("Total attention heads:", total_heads)
```

### 3. Extract the third layer and second head

Python indices start at zero, so the third layer uses index `2` and the second head uses index `1`. Each layer's attention tensor has shape `(batch_size, number_of_heads, sequence_length, sequence_length)`.


```python
bert_sentence = "Transformers use self-attention to understand language."
bert_inputs = bert_tokenizer(bert_sentence, return_tensors="pt")
bert_tokens = bert_tokenizer.convert_ids_to_tokens(
    bert_inputs["input_ids"][0].tolist()
)

with torch.no_grad():
    bert_outputs = bert_model(**bert_inputs, output_attentions=True)

# Select layer 3, batch item 1, and attention head 2.
# attentions[layer][batch, head, query_token, key_token]
third_layer_second_head = bert_outputs.attentions[2][0, 1]

print("Tokens:", bert_tokens)
print("Attention matrix shape:", third_layer_second_head.shape)
print("Third layer, second attention head:\n", third_layer_second_head)

# Each row is a probability distribution over the input tokens.
print("Row sums:", third_layer_second_head.sum(dim=-1))
assert third_layer_second_head.shape == (len(bert_tokens), len(bert_tokens))
assert torch.allclose(
    third_layer_second_head.sum(dim=-1),
    torch.ones(len(bert_tokens)),
    atol=1e-5,
)
```

### 4. What the rows and columns represent

- **Rows are query tokens.** Each row shows how one token distributes its attention over the input tokens.
- **Columns are key tokens.** Each column identifies a token that can receive attention.
- The entry at row `i`, column `j` is the attention weight from token `i` to token `j` in the third layer's second head. A larger value means token `i` gives more attention to token `j` in this head.

The matrix includes `[CLS]`, `[SEP]`, punctuation, and any WordPiece subwords produced by the tokenizer. If the tokenized sequence contains `N` tokens, the matrix is `N × N`. For example, the row for `self` and column for `attention` show how much `self` attends to `attention`.

Each row sums to approximately 1 because attention uses softmax over the key tokens. `eval()` disables dropout, and `torch.no_grad()` avoids computing gradients during inference. Attention can be asymmetric, so the weight from token `i` to token `j` can differ from the reverse direction.
